### Case - Cancelamento de Clientes

Você foi contratado por uma empresa com mais de 800 mil clientes para um projeto de Dados. Recentemente a empresa percebeu que da sua base total de clientes, a maioria são clientes inativos, ou seja, que já cancelaram o serviço.

Precisando melhorar seus resultados ela quer conseguir entender os principais motivos desses cancelamentos e quais as ações mais eficientes para reduzir esse número.

```python
# Step 1: Import data
import pandas as pd

table = pd.read_csv("cancelamentos.csv")

# Step 2: Visualize data (Understanding informations and find problems)
display(table)
```

```python
# Step 3: Fix the problems

display(table.info())

# Problem 1: Useless
table = table.drop(columns="CustomerID")

# Problem 2: Null
table = table.dropna()

# Problem 3: Wrong format
display(table.info())
# display(table)
```

```python

# Step 4: Initial analysis (How much clients canceled)

print(table["cancelou"].value_counts())

print(table["cancelou"].value_counts(normalize=True).map("{:.1%}".format))
```

```python
# Step 5: Analyze the cause of cancellation

import plotly.express as px

for col in table.columns:
    graph = px.histogram(table, x=col, color="cancelou", text_auto=True)
    graph.show()

```

```python
# Simulating fixing problems

table = table[table["ligacoes_callcenter"] <= 5]
table = table[table["meses_ultima_interacao"] <= 15]
table = table[table["dias_atraso"] <= 20]
table = table[table["duracao_contrato"] != "Monthly"]

display(table["cancelou"].value_counts(normalize=True))
```
