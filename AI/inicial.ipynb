```python
# Step 0: Understand the problem and the company
# Step 1: Import the data

import pandas as pd

df = pd.read_csv("clientes.csv")

display(df)
```

```python
# Step 2: Prepare the database for the AI

from sklearn.preprocessing import LabelEncoder

# print(df.info())

print({a for a in df["profissao"]})

codifier_prof = LabelEncoder()
codifier_cred = LabelEncoder()
codifier_pag = LabelEncoder()

df["profissao"] = codifier_prof.fit_transform(df["profissao"])
df["mix_credito"] = codifier_cred.fit_transform(df["mix_credito"])
df["comportamento_pagamento"] = codifier_pag.fit_transform(df["comportamento_pagamento"])

print(df.info())

display(df)
```

```python
# Divide in X and Y
# X (use to predict) -> all other columns
# Y (predict) -> score_credito

y = df["score_credito"]
x = df.drop(columns=["score_credito", "id_cliente"])

# Split data into training and test sets
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, train_size=0.7)
```

```python
# Step 3: Create model -> predict credit score

from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier

# RandomForest
# forest_model = RandomForestClassifier()
# RandomForest upgraded with parameters
forest_model = RandomForestClassifier(n_estimators=200, random_state=42)
# KNN
# knn_model = KNeighborsClassifier()
# KNN upgraded with parameters
knn_model = KNeighborsClassifier(n_neighbors=5, weights='distance')

# training models
forest_model.fit(x_train, y_train)
knn_model.fit(x_train, y_train)
```

```python
# Step 4: Choose the best model
from sklearn.metrics import accuracy_score

# Predicts
predict_forest = forest_model.predict(x_test)
predict_knn = knn_model.predict(x_test)

# Accuracy
display(accuracy_score(y_test, predict_forest)) # Best for now with 82.66%
display(accuracy_score(y_test, predict_knn))    # 75.82%
```

```python
# Step 5: Make new predicts

# New clients
new_df = pd.read_csv("novos_clientes.csv")

new_df["profissao"] = codifier_prof.fit_transform(new_df["profissao"])
new_df["mix_credito"] = codifier_cred.fit_transform(new_df["mix_credito"])
new_df["comportamento_pagamento"] = codifier_pag.fit_transform(new_df["comportamento_pagamento"])

predict_forest = forest_model.predict(new_df)
predict_knn = knn_model.predict(new_df)

display(predict_forest)
display(predict_knn)

display(new_df)
```
